### 01. Encapsulation
- Definition: Encapsulation is the process of wrapping data (attributes) and methods into a single unit called a class. It also helps control access to the data.

##### Real-world example: Bank Account.
- There is a balance in a bank account. We manage this balance through methods such as deposits and withdrawals.

```python
# Creating a BankAccount class
class BankAccount:

    def __init__(self, balance):       # Constructor
        self.__balance = balance       # Creating a private attribute

    def deposit(self, amount):         # Method to deposit money
        self.__balance += amount       # Updating the balance

    def show_balance(self):            # Method to display balance
        print("Balance:", self.__balance)


# Creating an object
account = BankAccount(5000)

# Depositing money
account.deposit(2000)

# Displaying the balance
account.show_balance()
```

### 2. Inheritance
- Definition: Inheritance is an OOP concept in which one class acquires the attributes and methods of another class.

##### Important Terms:
- Parent Class: The class whose properties are inherited.
- Child Class: The class that inherits properties from another class.

```python
# Creating a parent class
class Animal:

    def eat(self):                   # Parent class method
        print("Animal is eating")


# Creating a child class
class Dog(Animal):

    def bark(self):                  # Child class method
        print("Dog is barking")


# Creating an object of Dog class
d1 = Dog()

# Calling parent class method
d1.eat()

# Calling child class method
d1.bark()
```

#### Types of Inheritance
##### Python supports five commonly discussed types:
- 1. Single Inheritance:- One child class inherits from one parent class.
- 2. Multiple Inheritance:- One child class inherits from multiple parent classes.
- 3. Multilevel Inheritance:- A class inherits from another child class, forming a chain.
- 4. Hierarchical Inheritance:- Multiple child classes inherit from one parent class.
- 5. Hybrid Inheritance:- A combination of two or more inheritance types.

### 3. Polymorphism
- Definition: Polymorphism is an OOP concept that allows the same method or interface to perform different behaviours depending on the object.

##### Real-world example: Different animals make different sounds.

```python
# Creating a Dog class
class Dog:

    def sound(self):                 # Dog sound method
        print("Dog barks")


# Creating a Cat class
class Cat:

    def sound(self):                 # Cat sound method
        print("Cat meows")


# Creating objects
d1 = Dog()
c1 = Cat()

# Calling the same method
d1.sound()
c1.sound()
```

##### Another Example – Method Overriding
- Definition: Method overriding occurs when a child class provides its own implementation of a method already defined in its parent class.

```python
# Creating a parent class
class Animal:

    def sound(self):
        print("Animal makes a sound")


# Creating a child class
class Dog(Animal):

    def sound(self):                 # Overriding parent method
        print("Dog barks")


# Creating an object
d1 = Dog()

# Calling the overridden method
d1.sound()
```

### 4. Abstraction
- Definition: Abstraction is an OOP concept that hides implementation details and shows only the essential features to the user.

##### Real-world example: ATM machine:
- User ATM par transaction perform karta hai without knowing its internal implementation.

```python
# Importing abstract class tools
from abc import ABC, abstractmethod


# Creating an abstract class
class Shape(ABC):

    @abstractmethod
    def area(self):                  # Declaring abstract method
        pass


# Creating a child class
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length         # Assigning length
        self.width = width           # Assigning width

    def area(self):                  # Implementing abstract method
        return self.length * self.width


# Creating an object
r1 = Rectangle(5, 4)

# Calling the area method
print("Area:", r1.area())
```
