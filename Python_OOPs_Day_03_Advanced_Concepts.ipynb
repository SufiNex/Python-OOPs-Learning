### 01. Instance Method
- Definition: An instance method is a method that works with the instance (object) of a class. It takes self as its first parameter and can access instance variables and class variables.

```python
class Student:

    def __init__(self, name, age):
        self.name = name  # Store student name
        self.age = age    # Store student age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

# Create an object
student1 = Student("Sufiyan", 22)

# Call the instance method
student1.display()
```

### 02. Class Method
- Definition: A class method is a method that works with the class itself rather than a specific object. It is defined using the @classmethod decorator and takes cls as its first parameter.

```python
class Student:

    college_name = "Radharaman Institute of Technology & Science (RITS)"  # Class variable

    @classmethod
    def show_college(cls):
        print("College Name:", cls.college_name)

# Call the class method
Student.show_college()
```

### 03. Static Method
- Definition: A static method is a method that belongs to a class but does not depend on instance or class variables. It is defined using the @staticmethod decorator.

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

# Call the static method
result = Calculator.add(10, 20)

print("Addition:", result)
```

### 04. super() Function
- Definition: The super() function is used to call methods or access functionality from a parent class inside a child class.

```python
class Animal:

    def sound(self):
        print("Animals make sounds")

class Dog(Animal):

    def sound(self):
        super().sound()  # Call parent class method
        print("Dog barks")

# Create an object
dog = Dog()

# Call the method
dog.sound()
```

### 05. Constructor Chaining
- Definition: Constructor chaining is the process of calling a parent class constructor from a child class constructor.

```python
class Person:

    def __init__(self, name):
        self.name = name  # Initialize name
        print("Person constructor called")

class Student(Person):

    def __init__(self, name, roll_no):
        super().__init__(name)  # Call parent constructor
        self.roll_no = roll_no  # Initialize roll number

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)

# Create an object
student1 = Student("Sufiyan", 101)

# Display student details
student1.display()
```

### 06. Magic Methods (Dunder Methods)
- Definition: Magic methods are special methods in Python that have double underscores before and after their names. They allow objects to work with built-in Python operations.

- Examples: __init__(), __str__(), __len__(), __add__().

#### 6.1 __init__() Method
- Definition: The __init__() method initializes the attributes of an object when it is created.

```python
class Student:

    def __init__(self, name):
        self.name = name

student1 = Student("Sufiyan")

print(student1.name)
```

#### 6.2 __str__() Method
- Definition: The __str__() method returns a readable string representation of an object.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student Name: {self.name}"

student1 = Student("Sufiyan")

print(student1)
```

#### 6.3 __len__() Method
- Definition: The __len__() method defines the behavior of the built-in len() function for an object.

```python
class Team:

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

team1 = Team(["Alisha", "Arif", "Sufiyan"])

print("Total Members:", len(team1))
```

#### 6.4 __add__() Method
- Definition: The __add__() method defines the behavior of the + operator for objects.

```python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

num1 = Number(10)
num2 = Number(20)

result = num1 + num2

print("Addition:", result)
```

### 07. Property Decorator (@property)
- Definition: The @property decorator allows a method to be accessed like an attribute. It is commonly used to control access to class data.

```python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

# Create an object
student1 = Student("Sufiyan")

# Access the property like an attribute
print("Student Name:", student1.name)
```
