# Python OOPs – Day 01
## Topics: Class, Object, Methods, Constructor & self

### 01. What is OOPs?
- Definition: Object-Oriented Programming (OOPs) is a programming paradigm based on classes and objects. It helps organize code, improve reusability, and represent real-world entities.

### 02. Class
- Definition: A class is a blueprint or template used to create objects. It defines the attributes and methods of objects.

##### A. Built-in Class
- Definition: Built-in classes are predefined classes provided by Python to perform specific operations.

```python
# Creating an integer object
a = 10
print(type(a))  # Displays the type of the object

# Creating a string object
name = "Sufiyan"
print(type(name))  # Displays the type of the object

# Creating a list object
marks = [80, 90, 95]
print(type(marks))  # Displays the type of the object

# Creating a dictionary object
student = {"name": "Alisha", "age": 21}
print(type(student))  # Displays the type of the object
```

##### B. User-defined Class
- Definition: A user-defined class is a class created by the programmer to define their own objects, attributes, and methods in Python.

```python
# Creating a user-defined class
class Student:
    pass  # Defines an empty class

# Creating an object
s1 = Student()  # Creates an object of the Student class

print(type(s1))  # Displays the type of the object
```

### 03. Object
- Definition: An object is an instance of a class that contains data (attributes) and behaviour (methods).

```python
# Creating a class
class Student:
    pass

# Creating objects
s1 = Student()  # Creates the first object
s2 = Student()  # Creates the second object

print(s1)  # Displays the first object
print(s2)  # Displays the second object
```

### 04. Function vs Method

##### A. Function
- Definition: A function is a reusable block of code that performs a specific task and is defined outside a class.

```python
# Defining a function
def greet():
    print("Hello, Welcome to Python")

# Calling the function
greet()
```

##### B. Method
- Definition: A method is a function defined inside a class that performs a specific task related to that class or its objects.

```python
# Creating a class
class Student:

    def study(self):  # Defining a method
        print("I am studying Python")

# Creating an object
s1 = Student()

# Calling the method using the object
s1.study()
```

### 05. Types of Methods

##### A. User-defined Method
- Definition: A user-defined method is a method created by the programmer inside a class to perform a specific task.

```python
# Creating a class
class Student:

    def display(self):  # Defining a user-defined method
        print("Welcome to Python OOPs")

# Creating an object
s1 = Student()

# Calling the method using the object
s1.display()
```

##### B. Special Method
- Definition: Special methods are predefined methods in Python that use double underscores before and after their names. They are automatically invoked during specific operations.

- Examples: __init__(), __str__(), __len__().

```python
# Creating a class
class Student:

    def __init__(self):  # Defining a special method
        print("Object created")

# Creating an object
s1 = Student()  # Automatically calls the __init__() method
```

### 06. Constructor – __init__()
- Definition: A constructor is a special method that is automatically called when an object is created. In Python, the __init__() method initializes the object's attributes.

```python
# Creating a class
class Student:

    def __init__(self):  # Defining a constructor
        print("Student object created")

# Creating objects
s1 = Student()  # Automatically calls the constructor
s2 = Student()  # Automatically calls the constructor again
```

### 07. What is self?
- Definition: The self parameter represents the current instance of a class. It is used to access the attributes and methods of the current object.

```python
# Creating a class
class Student:

    def display(self):  # self refers to the current object
        print("Hello Student")

# Creating objects
s1 = Student()
s2 = Student()

# Calling the method using objects
s1.display()  # self refers to s1
s2.display()  # self refers to s2
```

### 08. Method Without Argument
- Definition: A method without additional arguments is a method that accepts only the self parameter and does not require any extra arguments from the user.

```python
# Creating a class
class Student:

    def greet(self):  # Method without additional arguments
        print("Welcome Sufiyan")

# Creating an object
s1 = Student()

# Calling the method
s1.greet()
```

### 09. Method With Argument
- Definition: A method with arguments is a method that accepts additional parameters along with the self parameter to perform a specific task.

```python
# Creating a class
class Student:

    def greet(self, name):  # Method with an additional argument
        print("Welcome", name)

# Creating an object
s1 = Student()

# Calling the method with arguments
s1.greet("Sufiyan")
s1.greet("Alisha")
```

### 10. Constructor With Arguments
- Definition: A constructor with arguments is used to initialize an object with specific values when the object is created.

```python
# Creating a class
class Student:

    def __init__(self, name, age):  # Constructor with arguments
        self.name = name  # Assigning name to the object
        self.age = age    # Assigning age to the object

    def display(self):  # Defining a display method
        print("Name:", self.name)
        print("Age:", self.age)

# Creating an object with arguments
s1 = Student("Sufiyan", 21)

# Calling the display method
s1.display()
```
